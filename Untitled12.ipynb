````python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Let's generate a clean dataset specifically tailored for the Emergency Dispatch & Traffic Physics prompt
np.random.seed(42)
dates = pd.date_range(start='2023-01-01', periods=10000, freq='10Min')

data = {
    'timestamp': dates,
    'emergency_type': np.random.choice(['Medical', 'Fire', 'Traffic Accident', 'Assault'], size=10000),
    'dispatch_priority': np.random.choice(['Low', 'Medium', 'High', 'Critical'], size=10000),
    'response_time_min': np.random.normal(loc=12, scale=4, size=10000).clip(1, 60),
    'traffic_density': np.random.uniform(10, 100, size=10000),
    'average_speed': np.random.uniform(15, 80, size=10000),
    'weather_condition': np.random.choice(['Clear', 'Rainy', 'Foggy', 'Snowy'], size=10000),
    'congestion_index': np.random.uniform(0.1, 1.0, size=10000)
}

df = pd.DataFrame(data)

# Introduce 2% missing values to demonstrate cleaning
df.loc[df.sample(frac=0.02).index, 'traffic_density'] = np.nan
df.loc[df.sample(frac=0.01).index, 'weather_condition'] = np.nan

print("--- Task 1: Load and Inspect Dataset ---")
print(f"Initial shape: {df.shape}")
print(f"Missing values before fill:\n{df.isnull().sum()}\n")

# 1. Fill missing values
df['traffic_density'] = df['traffic_density'].fillna(df['traffic_density'].median())
df['weather_condition'] = df['weather_condition'].fillna(df['weather_condition'].mode()[0])

# 2. Convert categorical data into numbers
label_encoders = {}
categorical_cols = ['emergency_type', 'dispatch_priority', 'weather_condition']
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# 3. Normalize numerical values
numerical_cols = ['response_time_min', 'traffic_density', 'average_speed', 'congestion_index']
scaler = MinMaxScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print("Missing values after fill:\n", df.isnull().sum())
print("\nPreprocessed Dataset Sample:")
print(df.head())
````

````python
# --- Task 2: Arrange by time, group by hours, and create sequences ---

# Group by hour and aggregate with mean
df_hourly = df.set_index('timestamp').resample('H').mean().reset_index()
print(f"Hourly grouped dataset shape: {df_hourly.shape}")

# Create sequential data for deep learning model
def create_sequences(data, seq_length):
    X = []
    y = []
    # Using congestion_index as target (column index of target in df_hourly)
    target_idx = df_hourly.columns.get_loc('congestion_index') - 1 # subtracting 1 because we drop timestamp

    feature_data = data.drop(columns=['timestamp']).values
    for i in range(len(feature_data) - seq_length):
        X.append(feature_data[i:i+seq_length])
        y.append(feature_data[i+seq_length, target_idx])
    return np.array(X), np.array(y)

SEQUENCE_LENGTH = 12 # Look back 12 hours
X, y = create_sequences(df_hourly, SEQUENCE_LENGTH)
print(f"Features shape: {X.shape}, Targets shape: {y.shape}")
````

````python
# --- Task 3 & 4: Model selection, justification, and training ---
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Split into Train (80%) and Test (20%)
split = int(len(X) * 0.8)
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# Convert to PyTorch tensors
train_dataset = TensorDataset(torch.tensor(X_train, dtype=torch.float32), torch.tensor(y_train, dtype=torch.float32))
test_dataset = TensorDataset(torch.tensor(X_test, dtype=torch.float32), torch.tensor(y_test, dtype=torch.float32))

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# GRU model justification: Recurrent networks are perfect for sequential, hourly-aggregated telemetry data.
# GRU keeps historical dependencies with fewer parameters than LSTM.
class GRUPredictor(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim=1, num_layers=1):
        super(GRUPredictor, self).__init__()
        self.gru = nn.GRU(input_dim, hidden_dim, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        out, _ = self.gru(x)
        out = self.fc(out[:, -1, :])
        return out.squeeze(-1)

model = GRUPredictor(input_dim=X.shape[2], hidden_dim=32)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Train the model
model.train()
for epoch in range(15):
    epoch_loss = 0
    for batch_x, batch_y in train_loader:
        optimizer.zero_grad()
        predictions = model(batch_x)
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item() * len(batch_x)
    print(f"Epoch {epoch+1}/15 - Loss: {epoch_loss/len(X_train):.5f}")
````

### GRU Architecture Diagram & Flow

Below is the schematic representing how sequential features pass through the GRU and fully connected layers to predict future traffic congestion:

```
                      [ Input Sequence (12 Hours × 7 Features) ]
                                           │
                                           ▼
  ┌───────────────────────────────── GRU Cell ─────────────────────────────────┐
  │                                                                            │
  │       ┌─────────────────► [ Update Gate (z_t) ] ◄─────────────────┐        │
  │       │                                                           │        │
  │  [Prev Hidden State (h_t-1)]                              [Current Input (x_t)]
  │       │                                                           │        │
  │       └─────────────────►  [ Reset Gate (r_t) ]  ◄────────────────┘        │
  │                                    │                                       │
  │                                    ▼                                       │
  │                        [ Candidate State (h~_t) ]                          │
  │                                    │                                       │
  │                                    ▼                                       │
  │                        [ Final Hidden State (h_t) ]                        │
  └────────────────────────────────────┬───────────────────────────────────────┘
                                       │ (Output of last step of sequence)
                                       ▼
                         [ Fully Connected Layer (Linear) ]
                                       │
                                       ▼
                       [ Congestion Index Prediction (y_t) ]
```

````python
# --- Task 5: Plotting actual vs predicted and testing on unseen data ---
import matplotlib.pyplot as plt

model.eval()
predictions = []
with torch.no_grad():
    for batch_x, _ in test_loader:
        preds = model(batch_x)
        predictions.extend(preds.numpy())

predictions = np.array(predictions)

# Calculate metrics
mae = np.mean(np.abs(predictions - y_test))
rmse = np.sqrt(np.mean((predictions - y_test)**2))
print(f"\nPerformance Metrics on Unseen Test Data:")
print(f"MAE: {mae:.5f}")
print(f"RMSE: {rmse:.5f}")

# Plotting
plt.figure(figsize=(12, 6))
plt.plot(y_test[:100], label='Actual Congestion Index', color='blue', alpha=0.7)
plt.plot(predictions[:100], label='Predicted Congestion Index', color='orange', linestyle='--', alpha=0.9)
plt.title('Actual vs Predicted Traffic Congestion (First 100 timesteps of Unseen Test Data)')
plt.xlabel('Time (Hours)')
plt.ylabel('Normalized Congestion Index')
plt.legend()
plt.grid(True)
plt.show()
````
